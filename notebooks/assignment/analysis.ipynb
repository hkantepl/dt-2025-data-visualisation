```python
import pandas as pd

df= pd.read_csv("Sleep_health_and_lifestyle_dataset.csv")
df.head()
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
df.groupby("Age")["Quality of Sleep"].mean().plot(kind='bar', color='pink')
plt.title("Average Sleep Quality by age")
plt.xlabel("Age")
plt.ylabel("Average Sleep Quality")
# plt.show()
```

1. From the above table we can see that there is a big dip in sleep quailty in the age 34. This may come from the fact, that people around that age may have small kids while still tackling big work responsibilities
2. Starting from around the age 51, we can see that the sleep quality gets better. It could be that the individuals value sleep more, and try to take care of their routine by getting enough sleep

```python
import matplotlib.pyplot as plt

Occupation_per_count=df["Occupation"].value_counts()
plt.figure(figsize=(8, 8))
Occupation_per_count.plot(kind='pie', legend=False, autopct='%1.1f%%')
plt.title("Occupations")
plt.show
```

1. We're able to see that doctors and nurses make up 38.5% of the whole list of occupations in the dataset. 
2. Another big portion of workers come from engineering side, being the third most popular occupations
3. The smallest sections belong to: scientists, software engineers, sales representatives

```python
size_of_bubble = df["Sleep Duration"]*30
plt.figure(figsize=(10, 6))
plt.scatter(x=df["Daily Steps"],  y=df["Sleep Duration"],s=size_of_bubble, c=df["Quality of Sleep"])
plt.title("Sleep quality, duration and daily steps correlation")
plt.xlabel("Daily Steps")
plt.ylabel("Sleep Duration")
plt.show
```

1. On average people have 8000 steps have also slept longer and average in a better sleep altogether. Therefore can make an assumption that the people who move more get better sleep as their bodies are more tired and not only stressed from work
2. What surprises from the table, is that people who do 10000 steps sleep less and have an average sleep in the lower quality as well.
3. Altogether it looks like that there is not a big correlation between peoples daily steps and their sleep quality and/or length

```python
plt.figure(figsize=(8, 6))
df.boxplot(column=["Quality of Sleep"], by="BMI Category")
plt.show

plt.title("Sleep quality by BMI Category")
plt.suptitle("")
plt.xlabel("BMI Category")
plt.ylabel("Quality of Sleep")
```

1. Here we have a clear indicator, that when a persons BMI is above normal, such as obese and/or overweight, their quality of sleep is dropping heavily. 
2. Most people in normal weight have ranked their sleep between 7-8 points out of 10. 
3. The lowest score comes from the highest BMI, ranking around 4.5 points out of 10.
